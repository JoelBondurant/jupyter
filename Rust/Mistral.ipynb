```rust
use mistralrs::{TextModelBuilder, TextMessages, TextMessageRole};
```

```rust
let model = TextModelBuilder::new("Qwen/Qwen2.5-Coder-1.5B-Instruct".to_string())
    .with_isq(mistralrs::IsqType::Q8_0) 
    .build()
    .await
    .unwrap();
```

```rust
let messages = TextMessages::new()
    .add_message(TextMessageRole::System, "You are a SQL generator. Output ONLY raw SQL code. No markdown, no explanations, no triple backticks.")
    .add_message(TextMessageRole::User, "Assume a Person table exists, select everyone over the age of 35.");
```

```rust
let response = model.send_chat_request(messages).await.unwrap();
```

```rust
println!("{}", response.choices[0].message.content.as_ref().unwrap());
```
